Bài 3.1 – Soi lỗi đặt tên
Cho danh sách định danh sau, sinh viên xác định định danh nào hợp lệ, định danh nào sai và giải
thích lý do:
1diem gia-tri _tam_thoi Diem_TB
class so luong MAX_SPEED diemTB
2024_data tong$ sinhVien1

1. 1diem => bắt đầu bằng số => sai.
2. gia-triSai => chứa dấu gạch ngang (-)=> sai.
3. _tam_thoi: Hợp lệ
4. Diem_TB: Hợp lệ về cú pháp nhưng chưa đúng chuẩn PEP8.
5. class => Sai vì là từ khóa của Python.
6. so luong => chứa dấu cách => sai.
7. MAX_SPEED: Hợp lệ vì đúng kiểu đặt tên hằng số.
8. diemTB: Hợp lệ về cú pháp nhưng chưa đúng chuẩn PEP8.
9. 2024_data => bắt đầu bằng số => sai.
10. tong$ => chứa ký tự đặc biệt $ => sai.
11. sinhVien1: Hợp lệ về cú pháp nhưng chưa đúng chuẩn PEP8.

Câu 1: Sự khác biệt giữa chạy file .py và gõ trực tiếp trong REPL là gì?
File.py chứa toàn bộ chương trình và có thể lưu lại, chạy lại nhiều lần. Gõ trực tiếp trong REPL cho phép nhập và thực thi từng lệnh ngay lập tức, phù hợp để thử nghiệm nhanh.
Câu 2: Khi nào nên dùng Jupyter Notebook thay vì file.py thông thường?
Khi học Python thì nên dùng Jupyter Noebook, thử nghiệm code, phân tích dữ liệu, vẽ biểu đồ hoặc muốn kết hợp code + kết quả + phần giải thích trong cùng một file. File.py thì phù hợp hơn khi xây dựng chương trình hoàn chỉnh.

Hoạt động 4: Từ khóa (keyword)
Bài tập nhanh:
• Chạy đoạn code sau để liệt kê toàn bộ từ khóa của Python:
import keyword
print(keyword.kwlist)
print("Số lượng từ khóa:", len(keyword.kwlist))
• Thử đặt tên biến trùng với một từ khóa, ví dụ class = 5, quan sát lỗi SyntaxError mà Python trả về rồi
giải thích.
=> class là một keyword (từ khóa) được Python dành riêng để khai báo lớp, nên không thể sử dụng class làm tên biến.
• Tại sao True, False, None cũng là từ khóa chứ không phải định danh thông thường?
=> True, False và None là các giá trị đặc biệt được Python định nghĩa sẵn và có ý nghĩa cố định trong ngôn ngữ. True và False biểu diễn giá trị Boolean đúng và sai, còn None biểu diễn giá trị không có dữ liệu hoặc không có giá trị. Vì chúng có ý nghĩa đặc biệt và không thể tùy ý sử dụng làm tên biến nên chúng được coi là các từ khóa/giá trị dành riêng của Python.
